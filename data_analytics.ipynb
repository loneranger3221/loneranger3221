```python
class infoprint:
  def __init__(self,title,name,salary):
    self.job_title=title
    self.emp_name=name
    self.job_salary=salary
  def print_details(self):
    print(f"My job title is {self.job_title} and my employer name is {self.emp_name} and my salary is {self.job_salary}")
    return
```

```python
job_title=input("Enter the job title")
emp_name=input("Enter the employer name")
job_salary=input("Enter the job salary")
```

```python
ob=infoprint(job_title,emp_name,job_salary)
ob.print_details()
```
